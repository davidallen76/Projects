```python
#imports 
import os
from dotenv import load_dotenv
from scraper import fetch_website_contents
from IPython.display import Markdown, display
from openai import OpenAI
```

```python
# Loead environment variables from .env file

load_dotenv(override=True)
api_key = os.getenv('OPENAI_API_KEY')

#verify the key

if not api_key:
    print("No API key found")
elif not api_key.startswith("sk-proj-"):
    print("Invalid API key format")
elif api_key.strip() != api_key:
    print("An API key should not have leading or trailing whitespace")
else:
    print("API key is valid")

```

```python
# Define a system prompt

system_prompt = """ 
You are a helpful assistant that summarizes the content of a website. 
Respond in markdown. Do not wrap the markdown in a code block - respond with only the markdown content.
"""
```

```python
# Define a user prompt

user_prompt_prefix = """ 
Here are the contents of a website.
Provide a short summary of this website.
If it includes news or announcements, include a section titled "News and Announcements" in your summary.
"""
```

```python
def messages_for(website):
    return [
        {"role": "system", "content": system_prompt},
        {"role": "user", "content": user_prompt_prefix + website}
]
```

```python
# testing messages_for function with a sample website. This should return the json object with the system and user prompts.
#  *Delete this later*
ed = fetch_website_contents("https://edwarddonner.com")
messages_for(ed)
```

```python
# simple function to call OpenAI 

openai = OpenAI()

def summarize(url):
    website = fetch_website_contents(url)
    response = openai.chat.completions.create(
        model="gpt-6-luna",
        reasoning_effort = "none",
        messages = messages_for(website)
    )

    return response.choices[0].message.content
```

```python
#testing the summarize function with a sample website. This should return a summary of the website on a single line
summarize("https://edwarddonner.com")
```

```python
# function to dsiplay the output of the summarize function in markdown format

def display_summary(url):
    summary = summarize(url)
    display(Markdown(summary))
```

```python
# testing the display_summary function.

display_summary("https://edwarddonner.com")
```
